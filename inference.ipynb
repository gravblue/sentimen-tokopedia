# Inference/Testing

```python
# Import Library
import numpy as np
import joblib
import tensorflow as tf
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.preprocessing.text import tokenizer_from_json
import json
```

```python
# Label Mapping
label_map = {0: "negative", 1: "neutral", 2: "positive"}

# Load Models
gru_model = tf.keras.models.load_model("gru_model.h5", compile=False)
svm_model = joblib.load("svm_model.pkl")
rf_model = joblib.load("random_forest_model.pkl")
lr_model = joblib.load("logistic_regression_model.pkl")

# Vectorizer TF-IDF
vectorizer = joblib.load("tfidf_vectorizer.pkl")

# Tokenizer Keras
with open("tokenizer.json", "r", encoding="utf-8") as f:
    tokenizer_json = f.read()
tokenizer = tokenizer_from_json(tokenizer_json)

# Contoh testing review Tokopedia
sample_reviews = [
    "Barang sesuai deskripsí, pengiriman cepat!",
    "Pelayanan toko sangat buruk, saya kecewa.",
    "Fitur aplikasi biasa saja, tidak spesial."
]

# Preprocessing
seq = tokenizer.texts_to_sequences(sample_reviews)
padded_seq = pad_sequences(seq, maxlen=80)
tfidf_vec = vectorizer.transform(sample_reviews)

# Prediction
pred_gru = np.argmax(gru_model.predict(padded_seq), axis=1)
pred_svm = svm_model.predict(tfidf_vec)
pred_rf = rf_model.predict(tfidf_vec)
pred_lr = lr_model.predict(tfidf_vec)

# Display Results
for i, text in enumerate(sample_reviews):
    print(f"Review {i+1}: {text}")
    print(f"GRU Model           : {label_map[pred_gru[i]]}")
    print(f"SVM Classifier      : {label_map[pred_svm[i]]}")
    print(f"Random Forest       : {label_map[pred_rf[i]]}")
    print(f"Logistic Regression : {label_map[pred_lr[i]]}\n")
```
